### Precision

Numbers and calculations are stored on computers to a specific level of precision. For example,

1. How many decimal places is the fraction two-thirds calculated to below?

````python
2/3
````

From what we know about calculations, we expect that $1 - \frac{2}{3} = \frac{1}{3} = 0.333\!\ldots$, with those decimal places continuing forever. However, in Python, that's not the case!

````python
1 - (2/3)
````

This is a specific type of rounding error that happens on computers, called **floating point error**. It's not that Python can't calculate the value (see below), it's that the way it is being asked to calculate the value introduces some rounding error. There are ways to avoid introducing this type of rounding error, such as simplifying the calculation before we give it to Python.

````python
1/3  # since 1 - (2/3) = 1/3
````

2. How might we simplify this calculation to avoid introducing floating point error?

````python
(9/10) - (8/9)
````

The choice of when we round in a calculation is also very important. For example, from how we define multiplication we know that

$$
    \underbrace{\frac{1}{47} + \frac{1}{47} + \cdots + \frac{1}{47}}_{47~\text{times}} = 47\times\frac{1}{47} = 1.
$$

However, using Python's function `round`, we can see that $\frac{1}{47}$ rounded to three decimal places is

````python
round(1/47, 3)
````

If we add this value together 47 times, we would expect to get 1. However, that's not the case!

````python
47 * round(1/47, 3)
````

1. Calculate the **magnitude of the error**, _i.e._ the difference in absolute value between the expected value and the calculated value.
2. If we increase the number of decimal places we round to, what would we expect to happen to the magnitude of the error? Why?

This type of error is what we call **premature rounding**, and while it's easily avoidable on a computer (just don't `round`!) it's fairly easy to fall into the trap when working through a calculation on paper.
